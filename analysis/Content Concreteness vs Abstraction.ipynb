__Goal:__

Measuring abstractness vs. concreteness is as a proxy for stylistic sophistication. assumption: correlates with elitism, academic tone, or register.

WordNet info: https://web.stanford.edu/~jurafsky/slp3/old_oct19/C.pdf

```python
words = ["justice", "idea", "democracy", "truth", "food", "car", "tree", "apple"]
```

```python
from nltk.corpus import wordnet as wn

def wordnet_concreteness(word):
    synsets = wn.synsets(word, pos='n')
    if not synsets:
        return None
    # Use the maximum depth (deepest meaning)
    depths = [s.min_depth() for s in synsets]
    return max(depths)

for w in words:
    print(w, wordnet_concreteness(w)) # Higher depth = more concrete.
```

Each synset in WordNet is organized in a hierarchical taxonomy (a tree structure) that starts from a very general concept (like entity.n.01) and becomes more specific as you move down.

The depth of a synset is how far it is from the root node (entity.n.01).

.min_depth() returns the minimum number of “is-a” (hypernym) links between the synset and the root.

Relations Between Synsets

| Relation                           | Meaning                      | Example (using “dog.n.01”)  |
| ---------------------------------- | ---------------------------- | --------------------------- |
| **hypernyms()**                    | more general concept         | `animal.n.01`               |
| **hyponyms()**                     | more specific concept        | `puppy.n.01`, `lapdog.n.01` |
| **meronyms()**                     | parts of something           | `tail.n.01`, `paw.n.01`     |
| **holonyms()**                     | wholes that contain it       | `canine.n.02`               |
| **antonyms()**                     | opposite words (via lemma)   | `good` ↔ `evil`             |
| **entailments()**                  | logical implications (verbs) | `snore` entails `sleep`     |
| **derivationally_related_forms()** | morphological links          | `decide` ↔ `decision`       |


# Wordnet Explorer

## 1. Synsets (Synonym Sets)

Each synset represents one sense (meaning) of the word.

```python
from nltk.corpus import wordnet as wn

word = "bank"
synsets = wn.synsets(word)
print(synsets)
```

Each synset groups together synonyms for that specific meaning.

## 2. Definitions and Examples

Each synset has:

A definition (gloss) — what the meaning is

Example sentences showing usage

```python
for syn in wn.synsets("bank"):
    print(f"{syn.name()}: {syn.definition()}")
    print("Examples:", syn.examples())
```

## Lemmas (Word Forms)

A lemma is a canonical form of a word within a synset (basically, a synonym in that sense).

```python
syn = wn.synset("bank.n.01")
print(syn.lemmas())
print([lemma.name() for lemma in syn.lemmas()])
```

4. Relations Between Synsets

WordNet encodes a web of semantic relations, including:

| Relation                           | Meaning                      | Example (using “dog.n.01”)  |
| ---------------------------------- | ---------------------------- | --------------------------- |
| **hypernyms()**                    | more general concept         | `animal.n.01`               |
| **hyponyms()**                     | more specific concept        | `puppy.n.01`, `lapdog.n.01` |
| **meronyms()**                     | parts of something           | `tail.n.01`, `paw.n.01`     |
| **holonyms()**                     | wholes that contain it       | `canine.n.02`               |
| **antonyms()**                     | opposite words (via lemma)   | `good` ↔ `evil`             |
| **entailments()**                  | logical implications (verbs) | `snore` entails `sleep`     |
| **derivationally_related_forms()** | morphological links          | `decide` ↔ `decision`       |


```python
syn = wn.synset("dog.n.01")
print("Hypernyms:", syn.hypernyms())
print("Hyponyms:", syn.hyponyms())
```

## Lexical Relations (Word-level)

Between lemmas rather than meanings. For example, antonyms:

```python
lemma = wn.lemma('good.a.01.good')
print(lemma.antonyms())
```

# Alternative 1: Using Concreteness Score from Study

* Concreteness vs abstraction level of words: https://link.springer.com/article/10.3758/s13428-013-0403-5?utm_source=chatgpt.com


* Dataset: https://link.springer.com/article/10.3758/s13428-013-0403-5#MOESM1

```python
import pandas as pd

# Example using Brysbaert et al. 2014 (download CSV first)
concreteness = pd.read_excel("concreteness_ratings_Brysbaert_et_al_2014.xlsx")

def concreteness_score(word):
    score = concreteness.loc[
        concreteness['Word'].str.lower() == word.lower(), 'Conc.M'
    ].values

    if len(score):
        print(f"Concreteness score for '{word}': {score[0]:.2f} (1=abstract, 5=concrete)")
    else:
        print(f"'{word}' not found in the dataset.")
```

```python
for w in words:
    concreteness_score(w)
```
